```python
np.random.seed(42)
n = 200

data = {
    "EmployeeID": np.arange(1, n+1),
    "Department": np.random.choice(["Sales", "IT", "HR", "Marketing"], n),
    "Age": np.random.randint(21, 60, n),
    "Salary": np.random.randint(35000, 95000, n),
    "ExperienceYears": np.random.randint(0, 20, n),
    "MonthlySales": np.random.randint(2000, 20000, n)
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
```

```python
# 9) CORRELATION
print(df.corr(numeric_only=True))
```

```python
# 10) CREATE SALARY CATEGORY
bins = [0, 40000, 60000, 80000, 100000]
labels = ["Low", "Medium", "High", "Very High"]
df["SalaryLevel"] = pd.cut(df["Salary"], bins=bins, labels=labels)
```

```python
# 11) MERGE
departments = pd.DataFrame({
    "Department": ["Sales","IT","HR","Marketing"],
    "Manager": ["Alice","Bob","Carol","David"]
})

merged_df = pd.merge(df, departments, on="Department")
print(merged_df.head())
```
